Une valeur manquante est-elle automatiquement une erreur métier?

Non, une valeur manquante n'est pas automatiquement une erreur métier. Par exemple, l'absence de CustomerID signifie que le
client n'est pas identifié dans l'enregistrement concerné. Cependant, cette absence ne prouve pas que la vente est
incorrecte. Il faut analyser le contexte et documenter cette limite avant toute décision de préparation des données.


Contrôle de traçabilité avec Python

Les statistiques descriptives montrent que la variable Quantity présente une moyenne de 9,55 et une médiane de 3, tandis que UnitPrice présente une moyenne de 4,61 et une médiane de 2,08. Les deux variables comportent des valeurs minimales négatives ainsi que des valeurs maximales élevées. Ces observations nécessitent une attention particulière lors de l'interprétation des données. À ce stade, aucune valeur n'est supprimée ou modifiée, car le TP1 porte sur le cadrage et la découverte des sources.

# Interprétation des types de données

### Analyse des types de données

L'inspection des types montre que le fichier contient quatre colonnes de type object, une colonne entière, deux colonnes de type float64 et une colonne de type datetime64[ns]. La colonne InvoiceDate permet d'identifier la date et l'heure des enregistrements. Quantity représente une quantité entière, tandis que UnitPrice contient des valeurs numériques décimales.

La colonne CustomerID est de type float64 et contient des valeurs manquantes. Ces caractéristiques sont à documenter avant toute opération de préparation des données.


Élément et Résultat observé

```python

print("Quantités négatives :",
      (df["Quantity"] < 0).sum())

print("Prix unitaires négatifs :",
      (df["UnitPrice"] < 0).sum())

print("Date minimale :",
      df["InvoiceDate"].min())

print("Date maximale :",
      df["InvoiceDate"].max())
```

```python
questions = {
    "Evolution temporelle": [
        "InvoiceDate", "Quantity", "UnitPrice"
    ],
    "Analyse des produits": [
        "StockCode", "Description", "Quantity"
    ],
    "Analyse par pays": [
        "Country", "Quantity"
    ],
    "Commandes et clients": [
        "InvoiceNo", "CustomerID"
    ]
}

for question, champs in questions.items():
    disponibles = all(champ in df.columns for champ in champs)

    print(question, ":", "Faisable" if disponibles
          else "Champs manquants")

    print("Champs nécessaires :", champs)
```
